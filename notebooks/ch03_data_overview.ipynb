## 40. 데이터 불러오기

```python
import pandas as pd

# pd(판다스).read_csv(파일경로(여기서도 파일이름 /로 이어붙이기))  -> 네 변수는 파이썬이 이해할 수 있는 매개변수 : df(데이터프레임)
customers = pd.read_csv(data_dir / "customers.csv") # 고객정보
products = pd.read_csv(data_dir / "products.csv") # 제품 정보
orders = pd.read_csv(data_dir / "orders.csv") # 주문 정보(영수증)
order_items = pd.read_csv(data_dir / "order_items.csv") # 세부 주문 내역
```

## 41. DataFrame 사전 만들기

```python

# 키-값 dictionary 만들기
datasets = {
    "customers": customers,
    "products": products,
    "orders": orders,
    "order_items": order_items,
}

print(datasets)
```

## 42. 크기 요약

```python
'''
pandas.DataFrame()은 파이썬 데이터 분석 라이브러리인 Pandas에서 '표(Table) 형태의 데이터 구조'를 만들어내는 클래스(함수)입니다.
엑셀의 시트(Sheet)나 SQL의 테이블처럼 행(Row)과 열(Column)로 이루어진 2차원 데이터 구조를 생성하는 기능을 합니다.

data (핵심 매개변수):
표에 들어갈 실제 데이터를 받습니다.
리스트, 딕셔너리, NumPy 배열 등 다양한 자료형을 전달할 수 있습니다.

columns (선택 매개변수):
각 열(Column)의 이름을 직접 지정할 때 사용합니다.
데이터가 딕셔너리 형태일 경우, 키(Key) 값이 자동으로 열 이름으로 채워집니다.

index (선택 매개변수):
각 행(Row)의 이름(인덱스)을 지정합니다. 기본값은 0, 1, 2... 형태의 정수입니다.
'''
shape_summary = pd.DataFrame(
    [
        # for문 리스트 컴프리핸션 {만들고 싶은 결과} for ~~
        {
            "dataset": name,
            #df.shape는 데이터프레임의 크기를 (행 개수, 열 개수) 형태의 튜플(Tuple)로 반환
            "rows": df.shape[0], #df.shape[0]: 인덱스 0번에 위치한 행(Row)의 개수
            "columns": df.shape[1], # df.shape[1]: 인덱스 1번에 위치한 열(Column)의 개수
            
        }
        
        for name, df in datasets.items()
    ]
)

print(shape_summary) #notebook에서는 shape_summary만써도 출력이 되지만 py파일에서는 안되므로 가급적 print문 사용하기
```

## 48. dtype 요약표

```python
dtype_rows = [] # dtype_rows를 가져올 수 있는 빈 리스트 변수 선언

for name, df in datasets.items(): # 4번 돌 것
    # 데이터 타입 수만큼 for문 돌 것(각 데이터프레임마다 다름)
    for column, dtype in df.dtypes.items(): # dtypes는 series객체 형태다. items()를 쓰지 않으면 컬럼명은 안나오고 데이터타입만 나옴 items는 쌍으로 뽑아내기 위함
        
        dtype_rows.append(
            {
                "dataset": name,
                "column": column,
                "dtype": str(dtype), #dtype은 파이썬 기본 문자열(str)이 아니라 NumPy나 Pandas의 전용 타입 객체(Object)이기 때문에 str()
            }
            '''
            딕셔너리 리스트 dtype_rows의 최종 형태 예시
            {"dataset": "customers", "column": "age", "dtype": "int64"},
            {"dataset": "customers", "column": "name", "dtype": "object"}
            이런 형태의 데이터 구조는 나중에 pd.DataFrame(dtype_rows)로 변환하여 깔끔한 표(데이터프레임)로 새로 만들 때 자주 사용
            '''
        )

dtype_summary = pd.DataFrame(dtype_rows)
print(dtype_summary)
# customers의 signup_date와 orders의 order_date는 날짜 형태임에도 str이므로 변환이 필요함을 알 수 있다.
```

## 49. 날짜형 후보 확인

```python
date_columns = {
    # 날짜 프레임이 들어있던 것들
    "customers": ["signup_date"], # 이 예제에서는 아니지만, 혹시 여러 개 일수 있으니 리스트로.
    "orders": ["order_date"],
}
# str형식을 날짜 형식으로 바꾸어야 함
for dataset_name, columns in date_columns.items():
    df = datasets[dataset_name]
    for column in columns:
        converted = pd.to_datetime( # 날짜 형식으로 바꾸는 메소드
            df[column],
            errors="coerce"
            '''
            errors="coerce"
            날짜로 바꾸기 힘든 이상한 값(예: "2024-13-45", "알수없음", "-" 등)을 만나면 오류를 무시하고
            NaT (Null/결측치)로 채웁니다.
            데이터 전처리 과정에서 어떤 행에 이상한 날짜 데이터가 섞여 있는지 찾아낼 때 핵심적으로 사용됩니다.
            errors="raise" (기본값)
            날짜 변환이 불가능한 값이 하나라도 있으면 즉시 ValueError 에러를 발생시키고 전체 코드가 중단됩니다.
            errors="ignore"
            변환할 수 없는 값을 만나면 에러를 내는 대신 변환하지 않고 원래 값(문자열 그대로)을 유지합니다.
            '''
        )

        print(
            dataset_name,
            column,
            "원본 dtype:", df[column].dtype,
            "변환 실패:", converted.isna().sum() # 결측치 세기
        )
```
